#**Python Part I: Data Wrangling & Cleaning**

First, some definitions. What is data cleaning and what is data wrangling?

**<font color="red">Data cleaning</font>** describes the specific process of fixing errors, handling null values and outliers, dealing with inconsistent data types.

**<font color="red">Data wrangling</font>** more broadly describes the beginning to end process of turning raw data into a useable form for data analysis and visualization. Wrangling includes data cleaning but might also involve normalization into a schema with primary and foreign keys, joining additional datasets, or pivoting/unpivoting columns and rows for easier analysis.

(e.g., If you receive an earnings report with Q1, Q2, Q3, Q4 in different columns, you may wish to unpivot that data (put all quarterly numbers in a single column) for easier analysis and then once your analysis is done, pivot the data again (Q1, Q2, Q3, Q4 spread out in separate columns) for visualization and presentation).

First, let's handle the items with a unique price: \$2.00 Coffee, \$1,50 Tea, \$5 Salad, and \$1.00 Cookie.

<details>
<summary><b>What command can we use to replace a null value with a replacement based on a matching value from in another column? (Hint: We need to pair the item and price, like a key and value pair)</b></summary>

<font color="red">def item_missing_cal(df):</font>

    # Creating Price map
    price_map = {2.0 : 'Coffee', 1.50 : 'Tea', 5.0 : 'Salad', 1.0 : 'Cookie'}

    # Create replacement using map
    replace = df['Price Per Unit'].map(price_map)

    # Calculate missing using replacement
    df['Item'] = df['Item'].fillna(replace)

<font color="red">item_missing_cal(df)</font>

</details>

<br>

Second, let's handle the items with the same \$3.00 Price: Cake and Juice. We want missing Items that correspond to \$3.00 to be replaced equally by Cake and Juice, so the data isn't unfairly skewed towards one or the other item.

<details>
<summary><b>What command can we use to replace a null value with a randomly sampled (50/50) replacement?</b></summary>

<font color="red">def missing_item_price_3(df):</font>

    # Item are at price == 3.0
    mapping = ['Cake', 'Juice']

    # Find rows that meet criteria
    masking = ((df['Item'].isnull()) & (df['Price Per Unit'] == 3))

    # Value Count of missing
    count = masking.sum()

    # Calculate missing
    if count > 0:
        df.loc[masking, 'Item'] = np.random.choice(mapping, size = count, p = [0.5, 0.5])                   # Here p = [0.5, 0.5] means it will fill 50%-50% both values.

<font color="red">missing_item_price_3(df)</font>

</details>

<br>

Finally, let's handle the items with the same \$4.00 Price: Sandwich and Smoothie. We want missing Items that correspond to \$4.00 to be replaced equally by Sandwich and Smoothie, so the data isn't unfairly skewed towards one or the other item.

<details>
<summary><b>How do we modify the previous command to do this?</b></summary>

<font color="red">def missing_item_price_4(df):</font>

    # Items at price 4
    items = ['Sandwich', 'Smoothie']

    # Find rows that meet criteria
    rows = ((df['Item'].isnull()) & (df['Price Per Unit'] == 4.0))

    # Count of these rows
    counts = rows.sum()

    # Calculate missing
    if counts > 0:
        df.loc[rows, 'Item'] = np.random.choice(items, size = counts, p = [0.5, 0.5])

Remember from our Exploratory Data Analysis, 33% of our Location column is null. A huge amount of the Payment Method is also null. If we drop the nulls we lose more than 33% of our whole dataset. However, we also don't have information from other columns to fill in these rows.

<details>
<summary><b>What do we do with the missing values in these two columns? </b></summary>

There is not enough information from other columns or sources to extrapolate these values, so we will have to replace the null values with randomly sampled data from the same column.

\# .isnull() creates a Boolean that identifies which rows have null values and .sample takes a random value from the same column and assigns it to the null row.
```

mask_pm = df['Payment Method'].isnull()

n_pm = mask_pm.sum()

if n_pm > 0:
  df.loc[mask_pm, 'Payment Method'] = (
      df['Payment Method'].dropna().sample(n=n_pm, replace=True).values
  )
```
\# This does the same for the Location column

```
mask_loc = df['Location'].isnull()

n_loc = mask_loc.sum()

if n_loc > 0:
  df.loc[mask_loc, 'Location'] = (
      df['Location'].dropna().sample(n=n_loc, replace=True).values
  )
```

</details>
